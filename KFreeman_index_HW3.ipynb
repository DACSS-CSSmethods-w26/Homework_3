# Part 1: Maximization/Minimization

Please, go to your _environment_ in Anacoda Navigator to install **glpk** and **pulp**  before runing the codes below.
Then, call the library:

```python
import pulp as pp
```

1. **Initialize the MODEL**: just write the name and declare if it is maximization or minimization problem type.

```python
model = pp.LpProblem(name='vitamin-supplements-problem', # just the name
                     sense=pp.LpMinimize) # other option: LpMinimize
```

2. **Declare the VARIABLES**: The refinery model consists of these _variables_:

```python
# how many SuperVit?
SV = pp.LpVariable(name="SuperVit",  # just the name
                    lowBound=0,  # ensure non-negativity
                    cat='Integer') # when would 'Integer' be needed?

# how many NewHealth?
NH = pp.LpVariable(name="NewHealth",
                 lowBound=0,
                 cat='Integer')
```

3. **Create function to OPTIMIZE**: The function is just the linear combination of the variables and their _given coefficients_:

```python
SVCoeff= 0.2
NHCoeff= 0.3
obj_func = SVCoeff*SV + NHCoeff*NH
```

4. **Represent the constraints**: These are the rules the model (set of variables) must obey:

```python
# SUBJECT TO:
C1= pp.LpConstraint(name='VITAMINC',   # just the name
                    e= 20*SV + 30*NH,
                    rhs=60,
                    sense=pp.LpConstraintGE) # >= 'sense'
C2= pp.LpConstraint(name='CALCIUM',
                    e= 500 * SV + 250 * NH,
                    rhs=1000,
                    sense=pp.LpConstraintGE) # >= 'sense'
C3= pp.LpConstraint(name='IRON',
                    e= 9*SV + 2*NH,
                    rhs=18,
                    sense=pp.LpConstraintGE, )# >= 'sense'
C4= pp.LpConstraint(name='NIACIN',
                    e= 2*SV + 10*NH,
                    rhs=20,
                    sense=pp.LpConstraintGE, )# >= 'sense'
C5= pp.LpConstraint(name='MAGNESIUM',
                    e= 60*SV + 90*NH,
                    rhs=360,
                    sense=pp.LpConstraintGE, )# >= 'sense'
```

5. **Build MODEL**: Here you add (i) the objective function, and (ii) all the constraints:

```python
model += obj_func
model += C1
model += C2
model += C3
model += C4
model += C5
```

6. **Solve the MODEL**: Notice we are not using the _default solver_:

```python
model.solve();
```

7. **Basic Report**:

From above, you can print:

```python
"Model Status",pp.LpStatus[model.status]
```

Also, get these key results:

```python
Results={"Optimal Solution to minimize cost":pp.value(model.objective)}
Results.update({v.name: v.varValue for v in model.variables()})
Results
```

From above, we know that 2 NewHealth tablets and 3 SuperVit tablets should be used in order to meet the nutrient requirements while minimizing costs. Since there is a requirement that the solutions are integers, a sensitivity analysis will not run in Excel and will not be used here.
